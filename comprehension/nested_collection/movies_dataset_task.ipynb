```python
# TITLES OF ALL MALAYALAM MOVIE

malayalam_movies = [m[0]for m in movies_data if m[1]== "malayalam"]
malayalam_movies
```

```python
#TITLES OF MOVIES RELEASED AFTER 2025

movies_after_2025 = [m[0]for m in movies_data if m[3] > 2015]
movies_after_2025
```

```python
# COUNT OF MOVIES HAVE A RATING OF 8.5 OR HIGHER

count =  len([m for m in movies_data if m[4]>=8.5])
count
```

```python
# AVERAGE DURATION

average =sum(m[2] for m in movies_data)/len(movies_data)
average

```
