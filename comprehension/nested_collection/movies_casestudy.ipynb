Movie Casestudy

```python
movies = [["kgf","kannada",150,2007,8],
          ["balan","malayalam",146,2026,7],
          ["aashique2","hindi",130,2007,9],
          ["bigb","malayalam",150,2004,9],
          ["drishyam","malayalam",145,2005,8]
          ]
          
    
```

```python
# movie title

all_move_titles = [m[0] for m in movies]
all_move_titles
```

```python
# all languages

all_languages = {m[1] for m in movies}
all_languages
```

```python
# all years

all_years = {m[3] for m in movies}
all_years
```

```python
# movie titles in uppercase

upper_case = [m[0].upper() for m in movies]
upper_case
```

```python
## movie titles starting with a

name_filter = [m[0] for m in movies if m[0].startswith("a")]
name_filter
```

```python
# display malayalam movies

malayalam_movies = [m[0] for m in movies if m[1]== "malayalam"]
malayalam_movies
```

```python
# display language wise count

languages = [m[1] for m in movies]

lc = {l:languages.count(l) for l in languages}
lc
```

```python
# plot


import matplotlib.pyplot as plt
x = lc.keys()
y = lc.values()
plt.bar(x,y)
```

```python
# highest run time

highest_run_time = [m[2] for m in movies]
highest = max(highest_run_time)
highest
```

```python
# lowest 

lowest_run_time = [m[2] for m in movies]
lowest = min(lowest_run_time)
lowest
```

```python
# average of runtime

average =sum(m[2] for m in movies)/len(movies)
average

```

```python
# latest movie

latest_year = max([m[3] for m in movies])
latest_movie = [m[0] for m in movies if m[3]==latest_year]
latest_movie
```

```python
# oldest movie

oldest_year = min([m[3] for m in movies])
oldest_movie = [m[0] for m in movies if m[3]==oldest_year]
oldest_movie

```

```python
# longest movie
run_time = max([m[2] for m in movies])
longest_movie = [m[0] for m in movies if m[2]==run_time]
longest_movie

```

```python
# shortest movie

run_time = min([m[2] for m in movies])
shortest_movie = [m[0] for m in movies if m[2]==run_time]
shortest_movie

```

```python
# year count

year = [m[3] for m in movies]

yc = {y:year.count(y) for y in year}
yc
```

```python
import matplotlib.pyplot as plt
x = yc.keys()
y = yc.values()
plt.bar(x,y)
```

```python
# highest rated movie

highest_rating = max([m[4] for m in movies])
highest_rated_movie = [m[0] for m in movies if m[4]==highest_rating]
highest_rated_movie
```

```python
#lowest rated movie

lowest_rating = min([m[4] for m in movies])
lowest_rated_movie = [m[0] for m in movies if m[4]==lowest_rating]
lowest_rated_movie

```

```python
# rating count

rating = [m[4] for m in movies]

rc = {r:rating.count(r) for r in rating}
rc
```

```python
#rating count

import matplotlib.pyplot as plt
x = rc.keys()
y = rc.values()
plt.bar(x,y)
```
