```python
books_data = [["The Hobbit", "Fantasy", 15, 1937, 4.8], # 0
             ["Dune", "Sci-Fi", 18, 1965, 4.7], # 1
             ["Project Hail Mary", "Sci-Fi", 22, 2021, 4.9], # 2
             ["Atomic Habits", "Self-Help", 16, 2018, 4.8], # 3
             ["1984", "Dystopian", 12, 1949, 4.6], # 4
             ["The Silent Patient", "Mystery", 14, 2019, 4.5], # 5
             ["Sapiens", "History", 20, 2014, 4.7], # 6
             ["Harry Potter", "Fantasy", 25, 1997, 4.9], # 7
             ["Educated", "Memoir", 17, 2018, 4.4], # 8
             ["Klara and the Sun", "Sci-Fi", 13, 2021, 4.2]] # 9
```

```python
# titles of all books more than 10 charactr

titles = [b[0] for b in books_data if len(b[0]) > 10]
titles
```

```python
# starts with the word "the"

books = [b[0] for b in books_data if b[0].startswith("The")]
books
```

```python
# unique genres

genres = {b[1] for b in books_data}
genres
```

```python
# count of books belong to sci fic

books = len([b for b in books_data if b[1] == "Sci-Fi"])
books
```

```python
# cost of all books

total = sum([b[2] for b in books_data])
total
```
