```python
# lambda function for adding two num

add_num = lambda n1,n2:n1+n2
print(add_num(100,500))
```

```python
# lambda function for subtracting two num

subtract_num = lambda n1,n2:n1-n2
print(subtract_num(40,30))
```

```python
# lambda function  for max of two num
max_two = lambda n1,n2:n1 if n1 > n2 else n2
max_two(100,200)
```

```python
# lambda function for min of two num

min_two = lambda n1,n2 : n1 if n1<n2 else n2
print(min_two(45,50))
```

```python
# user experience

user_experience = [
                   ["hari",6],
                   ["abin",8],
                   ["bijoy",4],
                   ["cibin",2],
                   ["akash",7],
                   ]



print(max(user_experience))

```

```python
# person with highest experience

print(max(user_experience,key = lambda lst:lst[1]))
```

```python
# person with lowest experience

print(min(user_experience,key = lambda lst:lst[1]))
```

```python
# weather

weather = [
           {"district":"tsr","temp":31},
           {"district":"pkd","temp":33},
           {"district":"knnur","temp":34},
           {"district":"kozhikkode","temp":38},
           {"district":"tvm","temp":37},
           {"district":"ekm","temp":36},
           ]
```

```python
# max

print(max(weather,key = lambda dictionary:dictionary.get("temp")))
```

```python
# min

min(weather,key = lambda dictionary:dictionary.get("temp"))
```
