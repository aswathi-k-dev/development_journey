```python
# level 3
# platform with highest total views


highest_views = {}
for p in viral_posts:
    platform = p.get("Platform")
    views = p.get("Views")
    if platform in highest_views:
        highest_views[platform]+=views
    else:
        highest_views[platform] = views
max(highest_views)





```

```python
# platform with highest total likess

highest_likes = {}
for p in viral_posts:
    platform = p.get("Platform")
    likes = p.get("Likes")
    if platform in highest_likes:
        highest_likes[platform]+=likes
    else:
        highest_likes[platform] = likes
max(highest_likes)
```

```python
# platform with highest total shares

highest_shares = {}
for p in viral_posts:
    platform = p.get("Platform")
    shares = p.get("Likes")
    if platform in highest_shares:
        highest_shares[platform]+=shares
    else:
        highest_shares[platform] = shares
max(highest_shares)
```

```python
# region  with highest total views

highest_views ={}
for p in viral_posts:
    region  = p.get("Region")
    views = p.get("Views")
    if region in highest_views:
        highest_views[region]+=views
    else:
        highest_views[region] = views
max(highest_views)
```

```python
# region  with highest total likes

highest_likes ={}
for p in viral_posts:
    region  = p.get("Region")
    likes= p.get("Likes")
    if region in highest_likes:
        highest_likes[region]+=likes
    else:
        highest_likes[region] = likes
max(highest_likes)
```

```python
# most commonly used hashtag

common_hashtag ={}
for p in viral_posts:
    hashtag = p.get("Hashtag")
    if hashtag in common_hashtag:
        common_hashtag[hashtag]+=hashtag
    else:
        common_hashtag[hashtag] = hashtag
max(common_hashtag)
```

```python
# commonly used content type

common_content ={}
for p in viral_posts:
    content = p.get("Content_Type")
    if content in common_content:
        common_content[content]+=content
    else:
        common_content[content] = content
max(common_content)
```

```python
# common engagemnt level

engagmnt_level = {}
for p in viral_posts:
    level = p.get("Engagement_Level")
    if level in engagmnt_level:
        engagmnt_level[level]+=level
    else:
        engagmnt_level[level] = level
max(engagmnt_level)
```
